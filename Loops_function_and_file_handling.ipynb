Map, reduce and **filter**

```python
#Area of circle by for loop
import math
def area(r):
  return math.pi*(r**2)
radii= [1,2,3,4,5]
calculated_areas=[]
for i in radii:
  calculated_areas.append(area(i))
print(calculated_areas)
```

```python
###map(f, iterable)
map(area, radii)
print(__builtins__.list(map(area, radii)))
```

Filter function

```python
import statistics
data= [1,2,3,4,5,6,7,8,9]
avg= statistics.mean(data)
avg
```

```python
filter(lambda x: x>avg, data)
```

```python
print(__builtins__.list(filter(lambda x: x>avg, data)))
```
