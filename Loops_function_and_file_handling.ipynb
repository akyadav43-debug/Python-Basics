Iterations

```python
#iterating over list
list =[10,20, 30, 40, 50]
for i in list:
  print(i)
```

```python
#iterating over string
str= "Anil"
for i in str:
  print(i)
```

```python
#iterating over Dictionary on keys

data= {1: ["Ram",23], 2:["Shyaam", 25]}
for i in data:
  print(i)
```

```python
#iterating over Dictionary on values
data= {1: ["Ram",23], 2:["Shyaam", 25]}
for i in data.items():
  print(i)
```

```python
#iterating over Dictionary on values
data= {1: ["Ram",23], 2:["Shyaam", 25]}
for i, j in data.items():
  print(j)
```
