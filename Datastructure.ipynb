Sets

1.   Sets are a type of collection like lists tuples, storing mixed data
2.   sets are enclosed wothin curly brackets and elements are written as comma seperated
sets are unordered collection of items
sets does not follow or allow duplicates



```python
a=[1,2,3,3,3,4,5,'A']
print(a)
b=set(a)
print(b)
print(type(a))
print(type(b))
print(len(a))
print(len(b))
```

```python
b
```

```python
b.add("Radha")
print(b)
b.remove("Radha")
print(b)
```

```python
Sets operations
Union, intersection
```

```python
#Union
A={1,2,3,5,3,4,7}
B={19,11,3,5,12,15}
print(A | B)
print(A.union(B))
print(len(A))
print(len(B))
print(len(A|B))
print(len(A | B))
```

```python
#Intersection
A={1,2,3,5,3,4,7}
B={19,11,3,5,12,15}
print(A & B)
print(A.intersection(B))
print(len(A & B))
print(len(A.intersection(B)))
```

```python
#symmetric difference\
A={1,2,3,5,3,4,7}
B={19,11,3,5,12,15}
print(A ^ B)
print(A.symmetric_difference(B))
print(len(A ^ B))
print(len(A.symmetric_difference(B)))
```

```python
#intersection of two list in sorted manner
A=[1,2,3,5,3,4,7]
B=[19,11,3,5,12,15]
list= (set(A) & set(B))
print(list)
print(sorted(list))
```

**Dictionary in python**



*   Unique and immutable
*   Values are accessed by their key
*   Values can be mutable, immutable and duplicats
*   Using Curly brackets
*   unordered collection of data
*   **Key value pairs**



```python
dict= {"name": "Anil", "Age": 26}
type(dict)
```

```python
#Accessing values by keys
dict["name"]
```

```python
dict["Age"]= 32
dict
```

```python
#inserting new key value pairs
dict["Sex"]= "Male"
dict
```

```python
#deleting a key value pair
del dict["Sex"]
dict
```

```python
#sorting dictionnary
sorted(dict)
```

```python
#Values() method
dict.values()
```

```python
#key() method
dict.keys()
```

```python
#get() method
dict.get("Age")
```

```python
#update() method
dict.update({"name": "Kapil"})
print(dict)
dict["name"]="Anil"
print(dict)
```
