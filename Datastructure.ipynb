***Sorting a tuple***
You can not directly sort tuple but to sort you have to first convert tuple in list then convert list to the tuple

```python
tup= (2,3,1,7,8,4,5,9)
newt= sorted(tup)
print(newt)
print(type(newt))
```

```python
#TYPE CASTING IN TUPLE
tup=tuple(newt)
print(tup)
print(type(tup))
```
