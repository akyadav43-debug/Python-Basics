```python
Dia=pd.read_csv('diabetes.csv')
```

```python
Dia.head(2)
```

```python
Dia.Pregnancies.value_counts().plot(kind='barh')
plt.show()
```

```python
Dia.Outcome.value_counts().plot(kind='barh')
```

```python
plt.bar(Dia['Pregnancies'], Dia['Outcome'].max(), color='Pink')
plt.title('Pregnancies vs Outcome')
plt.show()
```

```python
plt.scatter(Dia['Pregnancies'], Dia['Outcome'], color='Red')
plt.title('Pregnancies vs Outcome')
plt.xlabel('Pregnancies')
plt.ylabel('Outcome')
plt.show()
```

```python
plt.hist(Dia['Pregnancies'], bins=10, color='cyan', edgecolor='black')
plt.title('Pregnancies Histogram') 
plt.xlabel('Pregnancies')
plt.ylabel('Frequency')
plt.show()
```

```python
labels_ex= 'Java', 'Python', 'C++', 'C#', 'JavaScript'
sizes_ex= [100, 200, 300, 400, 500]
explode_labels= (1, 0, 0, 0, 0)
fig1, ax1= plt.subplots()
ax1.pie(sizes_ex, explode=explode_labels, labels=labels_ex, autopct='%1.1f%%', shadow=True, startangle=90)
plt.axis('equal')
plt.title('Programming Languages Popularity')
plt.show()
```

```python
Dia['Pregnancies'].plot.box()
plt.title('Pregnancies Box Plot')
plt.show()
```
