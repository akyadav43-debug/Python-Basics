```python
Data= {'Year':[2010,2011,2012,2013,2014,2015],
       'Sales':[1000,1500,2000,2500,3000,3500]}
Data
```

```python
type(Data)
```

```python
df
```

```python
type(df)
```

```python
df.plot(x='Year', y='Sales', kind='bar')
```

```python
df.plot(x='Year', y='Sales', kind='area')
```

```python
df.plot(x='Year', y='Sales', kind='barh')
```

```python
df.plot(x='Year', y='Sales', kind='scatter')
```
