Seaborn library

```python
import seaborn as sns
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('Diabetes.csv')
df.head()
```

```python
len(df)
```

```python
df.Outcome.value_counts().plot(kind='bar', color=['blue', 'orange'])
plt.title('Diabetes Outcome Count')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()
```

```python
sns.kdeplot(df['Pregnancies'], shade=True)
plt.title('Pregnancies Distribution')
plt.xlabel('Number of Pregnancies')
plt.ylabel('Density')
plt.show()
```

```python
sns.displot(df['Pregnancies'], kde=True, bins=10)
plt.title('Pregnancies Distribution')
plt.xlabel('Number of Pregnancies')
plt.ylabel('Count')
plt.show()
```

# KDE Plot Kernel Density Estimation (KDE) 
# It is a non-parametric way to estimate the probability density function of a random variable. 
# It is used to visualize the distribution of data points in a continuous manner.
