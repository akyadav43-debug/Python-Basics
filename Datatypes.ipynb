```python
#Extend and append function in list
list1=["Ram","shyam"]
list1.append([5,7])
print(list1)
len(list1)
```

```python
list1=["Ram","shyam"]
list1.extend([5,7])
print(list1)
len(list1)
```

```python
#Delete and pop operation in list works on index
list1=["Ram", "shyam", "Ramesh", 1,2, 3]
print(list1)
del list1[2]
print(list1)
```

```python
#pop operation in list works on index
list1=["Ram", "shyam", "Ramesh", 1,2, 3]
print(list1)
list1.pop(2)
print(list1)
```

```python
#remove operation in list works on element
list1=["Ram", "shyam", "Ramesh", 1,2, 3]
print(list1)
list1.remove("shyam")
print(list1)
```
