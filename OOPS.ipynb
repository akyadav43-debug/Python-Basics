Defining a class

```python
class rectangle :
    def __init__(self):
        self.length=10
        self.breadth=5
```

```python
rect=rectangle()
rect.length
```

```python
rect.breadth
```

PARAMETRISED Constructor

```python
class rectangle :
    def __init__(self):
        self.length=10
        self.breadth=5
```

```python
class rectangle :
    def __init__(self, length,breadth):
        self.length=length
        self.breadth=breadth
```

```python
rect=rectangle(5,10)
rect.length
```

```python
rect.breadth
```

CLASS VARIABLE AND INSTANCE VARIABLES

```python
class circle :
    pi=3.14
    def __init__(self,radius):
        self.radius=radius
```

```python
cicle1=circle(5)
print(cicle1.pi)
print(cicle1.radius)
```

```python
circle.pi=3
print(circle.pi)
```
