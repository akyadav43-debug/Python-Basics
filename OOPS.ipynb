PARAMETRISED Constructor

```python
class rectangle :
    def __init__(self):
        self.length=10
        self.breadth=5
```

```python
class rectangle :
    def __init__(self, length,breadth):
        self.length=length
        self.breadth=breadth
```

```python
rect=rectangle(5,10)
rect.length
```

```python
rect.breadth
```

CLASS VARIABLE AND INSTANCE VARIABLES

```python
class circle :
    pi=3.14
    def __init__(self,radius):
        self.radius=radius
```

```python
cicle1=circle(5)
print(cicle1.pi)
print(cicle1.radius)
```

```python
circle.pi=3
print(circle.pi)
```

Adding method to class

```python
class rectangle :
    def __init__(self, length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
##Instatiating the rectangle class
rect=rectangle(10,5)
```

```python
rect.length
```

```python
rect.breadth
```

```python
rect.area()
```

Inheritence and Overriding

```python
class Employee:
    def function1(self):
        print("This is function 1")

class department(Employee):
    pass
```

```python
emp=Employee()
emp.function1()
```

```python
class shape:
    def set_color(self,color):
        self.color=color
    def calculate_area(self):
        pass
    def color_the_shape(self):
        color_price= {"red":10,"blue":20,"green":30}
        return self.calculate_area()*color_price[self.color]
```

```python
class circle(shape):
    pi=3.14
    def __init__(self,radius):
        self.radius=radius
    def calculate_area(self):
        return circle.pi*self.radius*self.radius
```

```python
c=circle(5)
c.set_color("red")
print("circle with radius:",c.radius,"has area:",c.calculate_area(),"and color price:",c.color_the_shape())
```
